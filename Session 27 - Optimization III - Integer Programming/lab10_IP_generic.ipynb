## Simple example

Why is integer programming complex? Consider this simple example:

> $max \quad 21 X + 11 Y$

> subject to:

> $7 X + 4 Y <= 13$

> $X,Y >= 0	 \quad$ (Nonneg.)

**Idea**: Let's formulate it as a simple LP, then round up/down to get the integer rescrition.

```python
#Simple LP model & solution
model = pyo.ConcreteModel()
model.X = pyo.Var(bounds=(0, None))
model.Y = pyo.Var(bounds=(0, None))
X = model.X
Y = model.Y
model.obj = pyo.Objective(expr = 21*X + 11*Y, sense = maximize)
model.C1 = pyo.Constraint(expr = 7*X + 4*Y <= 13)
opt = SolverFactory('glpk')
results = opt.solve(model)
print("--------------- RESULTS --------------- \n");
print("Optimal objective function value =", model.obj());
print("X*", model.X())
print("Y*", model.Y())
```

If we are dealing with physical units, should we round $X^*$ up to $2$ or down to $1$?

Setting $X=2$ is not feasible as it violates the budget constraint  ($7*2=14 > 13$)!

Setting $X=1$ gives an OFV of $21$. This is feasible, but is it optimal?

```python
#Solving with integer variables
model = pyo.ConcreteModel()
model.X = pyo.Var(within=NonNegativeIntegers) #This is the only change needed
model.Y = pyo.Var(within=NonNegativeIntegers) #This is the only change needed
X = model.X
Y = model.Y
model.obj = pyo.Objective(expr = 21*X + 11*Y, sense = maximize)
model.C1 = pyo.Constraint(expr = 7*X + 4*Y <= 13)
opt = SolverFactory('glpk')
results = opt.solve(model)
print("--------------- RESULTS --------------- \n");
print("Optimal objective function value =", model.obj());
print("X*", model.X())
print("Y*", model.Y())
```

As we can see from the optimization output, the optimal solution is actually $X^*=0$ and $Y^*=3$ yielding an OFV of $33$! $\Rightarrow$ Integer programming solutions are not always easily predictable from the LP equivalent!
